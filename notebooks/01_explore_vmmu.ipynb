# Khám phá dataset VMMU

Tải và kiểm tra schema của dataset `anvo25/vmmu` (split `full_vqa`) từ HuggingFace Hub, để đánh giá xem có dùng được làm nguồn tham khảo/pretrain cho Đồ án 2 (hỏi đáp tài liệu tiếng Việt) không.

- Dung lượng: ~1.7GB
- Lưu ý: nếu chạy trên Colab, cache HuggingFace mặc định nằm ở runtime tạm và sẽ mất khi disconnect. Nếu cần chạy lại nhiều lần, set `HF_HOME`/`cache_dir` trỏ vào Google Drive để khỏi tải lại.

```python
import os

# Tăng số lần retry và timeout để tránh lỗi "connection forcibly closed" khi mạng không ổn định.
# Nếu có HF_TOKEN (đăng nhập HuggingFace), set vào .env để tăng rate limit và giảm khả năng bị ngắt kết nối.
os.environ.setdefault("HF_HUB_DOWNLOAD_TIMEOUT", "60")
os.environ.setdefault("HF_HUB_ETAG_TIMEOUT", "30")
```

```python
from datasets import load_dataset

vmmu = load_dataset("anvo25/vmmu", split="full_vqa")
vmmu
```

## Kiểm tra schema và một vài sample

```python
print("Features:", vmmu.features)
print("Num rows:", len(vmmu))
print()
print("Sample 0 (no image field shown):")
sample = vmmu[0]
for k, v in sample.items():
    if k.lower() != "image":
        print(f"- {k}: {v}")
```

```python
import matplotlib.pyplot as plt

n = 4
fig, axes = plt.subplots(1, n, figsize=(16, 4))
for i in range(n):
    row = vmmu[i]
    img = row.get("image")
    ax = axes[i]
    if img is not None:
        ax.imshow(img)
    ax.set_title(str(row.get("question", ""))[:60], fontsize=8)
    ax.axis("off")
plt.tight_layout()
plt.show()
```
