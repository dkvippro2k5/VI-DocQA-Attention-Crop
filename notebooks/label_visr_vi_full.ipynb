```python
# Xem kết quả (sample 5 items)
import json
from pathlib import Path

out_dir = Path('data/processed/visr_bench_vi')
labels_file = out_dir / 'qa_vi_haiku_full_batch_100.jsonl'

if labels_file.exists():
    with open(labels_file, encoding='utf-8') as f:
        items = [json.loads(line) for line in f]
    
    print(f'✓ Labeled {len(items)} items\n')
    
    # Show samples
    for item in items[:3]:
        print(f"QID: {item['qid']}")
        print(f"  Q: {item['question']}")
        print(f"  A: {item['answer']}")
        print(f"  Type: {item['question_type']}")
        print()
```
